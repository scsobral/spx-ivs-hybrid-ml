# Carregamento dos Dados

Este notebook tem como objetivo carregar todos os ficheiros CSV individuais da pasta `SPX_CBOE` (2597 ficheiros diários de 2009 a 2019), consolidá-los numa única tabela e guardar o resultado num formato eficiente para processamento futuro.

Data: 26/06/2025

Importação de bibliotecas

```python
import os
import pandas as pd
from datetime import datetime
from tqdm import tqdm
```

Designações centrais e caminhos

```python
DIR_INPUT = "02. Dados/01. Inputs/SPX_CBOE"
FICHEIRO_INPUT = None  # Não aplicável neste caso — múltiplos ficheiros.
FICHEIRO_OUTPUT = "02. Dados/01. Inputs/InputData_v0.01.parquet"
```

Carregamento de dados

```python
# Lista todos os ficheiros CSV no diretório
csv_files = sorted([f for f in os.listdir(DIR_INPUT) if f.endswith(".csv")])

# Carregar os dados com tqdm para progresso
df_total = pd.DataFrame()
for file in tqdm(csv_files, desc="Carregando ficheiros CSV"):
    df_temp = pd.read_csv(os.path.join(DIR_INPUT, file))
    df_total = pd.concat([df_total, df_temp], ignore_index=True)

# Verificações básicas
print("Shape final:", df_total.shape)
df_total["delivery_code"] = df_total["delivery_code"].astype(str)

df_total.head()
```

Guardar Dataframe

```python
# Guardar em formato parquet
import time

start = time.time()
df_total.to_parquet(FICHEIRO_OUTPUT)
print("Feather salvo em", round(time.time() - start, 2), "s")
```
