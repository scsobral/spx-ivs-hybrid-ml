Configuração Inicial

```python
import os

# Diretório base (usa "." para o diretório atual onde está o notebook)
BASE_DIR = "."

# Lista das subpastas a criar
pastas = [
    "01. Notebooks/00. Configuração/",
    "01. Notebooks/01. Exploração",
    "01. Notebooks/02. Transformações",
    "01. Notebooks/03. Modelação",
    "01. Notebooks/04. Resultados",
    
    "02. Dados/01. Inputs",
    "02. Dados/02. Dados Tratados",

    "03. Output/01. Gráficos",
    "03. Output/02. Tabelas",
    "03. Output/03. Excels",

    "04. Documentos/01. Versões Word",
    "04. Documentos/02. Imagens Word",
    "04. Documentos/03. Referências"
]

# Criar as pastas se não existirem
for pasta in pastas:
    path = os.path.join(BASE_DIR, pasta)
    os.makedirs(path, exist_ok=True)
    print(f"📁 Criada: {pasta}")

print("\n✅ Estrutura de pastas criada com sucesso.")
```
