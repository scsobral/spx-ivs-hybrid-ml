# Data Ingestion & Consolidation

This notebook loads all individual daily CBOE SPX option CSV files from 2009 to 2019 (2,597 trading day files), consolidates them into a single master table, and exports the raw consolidated dataset to Parquet format for downstream processing.

**Date:** June 2025


### 1. Library Imports


```python
import os
import pandas as pd
from datetime import datetime
from tqdm import tqdm
```

### 2. Path Definitions & Configuration


```python
DIR_INPUT = "02_data/raw/SPX_CBOE"
FICHEIRO_INPUT = None  # Multiple daily files
FICHEIRO_OUTPUT = "02_data/raw/InputData_v0.01.parquet"
```

### 3. Data Ingestion Loop


```python
# List all CSV files in the directory
csv_files = sorted([f for f in os.listdir(DIR_INPUT) if f.endswith(".csv")])

# Load data with tqdm progress bar
df_total = pd.DataFrame()
for file in tqdm(csv_files, desc="Loading daily CBOE CSV files"):
    df_temp = pd.read_csv(os.path.join(DIR_INPUT, file))
    df_total = pd.concat([df_total, df_temp], ignore_index=True)

# Basic schema verification
print("Final dataset shape:", df_total.shape)
df_total["delivery_code"] = df_total["delivery_code"].astype(str)

df_total.head()
```

### 4. Export Consolidated Dataset


```python
# Save figures em formato parquet
import time

start = time.time()
df_total.to_parquet(FICHEIRO_OUTPUT)
print("Parquet saved in", round(time.time() - start, 2), "s")
```
